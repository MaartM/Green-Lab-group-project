# 3. Run-level hypothesis tests (RQ1, RQ3, and η for RQ2)

The procedure of report Section 4.5, implemented in `stats.analyse` and applied to every
run-level variable:

1. **Check assumptions.** Shapiro-Wilk per strategy, median-centred Levene across strategies.
   If normality fails, the checks are repeated on log values.
2. **Omnibus test.** One-way ANOVA (equal variances) or Welch's ANOVA when normal,
   Kruskal-Wallis otherwise. α = 0.05.
3. **Pairwise comparisons, only when the omnibus test is significant.** Student's or Welch's
   t-test with Cohen's d, or Mann-Whitney U with Cliff's δ. Differences are in original
   units and as % of the second strategy.

Variables that are constant within a strategy are reported descriptively only.

```python
import pandas as pd
import matplotlib.pyplot as plt

import experiment_data as data
import stats
import style

style.apply()
pd.set_option("display.precision", 3)
runs = data.load_runs()
```

## 3.1 Assumptions and omnibus tests

```python
results = {variable: stats.analyse(runs, variable) for variable in data.RUN_VARIABLES}

def summary(results):
    rows = []
    for variable, r in results.items():
        rows.append({"variable": data.RUN_VARIABLES[variable],
                     "transform": r.get("transform", ""),
                     **{f"Shapiro p {s}": r.get("shapiro_p", {}).get(s) for s in data.STRATEGIES},
                     "Levene p": r.get("levene_p"), "test": r["test"], "p": r["p"]})
    return pd.DataFrame(rows).set_index("variable")

summary(results)
```

**Q-Q plots** for energy per query, the central RQ1 variable. Points on the line mean
normally distributed runs.

```python
from scipy.stats import probplot

fig, axes = plt.subplots(1, 3, figsize=(11, 3))
for ax, strategy in zip(axes, data.STRATEGIES):
    (theoretical, ordered), (slope, intercept, _) = probplot(
        runs.loc[runs["strategy"] == strategy, "energy_per_query_j"])
    ax.scatter(theoretical, ordered, s=12, color=style.COLORS[strategy])
    ax.plot(theoretical, slope * theoretical + intercept, color=style.MUTED, linewidth=1)
    ax.set_title(style.label(strategy))
    ax.set_xlabel("Theoretical quantile")
axes[0].set_ylabel("Energy per query (J)")
plt.tight_layout()
```

## 3.2 Pairwise comparisons

```python
pairwise = pd.concat({data.RUN_VARIABLES[v]: r["pairwise"]
                      for v, r in results.items() if r["pairwise"] is not None})
pairwise.droplevel(1).set_index("comparison", append=True)
```

## 3.3 Sensitivity: without the first run

The first run of the experiment is an outlier (notebook 2). It stays in the main analysis,
but repeating the tests without it shows whether it drives any conclusion. Most likely it
drives the failed normality checks of R.

```python
without_first = runs[runs["order"] > 0]
summary({v: stats.analyse(without_first, v) for v in data.RUN_VARIABLES})
```

## 3.4 Descriptive only: constant within a strategy

```python
constant = [v for v, r in results.items() if r["pairwise"] is None and pd.isna(r["p"])]
runs.groupby("strategy", observed=True)[constant].agg(["mean", "std"])
```

**Notes**

- **Every testable variable differs between strategies, and every pair differs.** This is
  expected with such small within-strategy spread.
- **The effect sizes are extreme.** Cliff's δ = ±1 means every run of one strategy lies
  above every run of the other. Cohen's d ranges from about 4 to over 300. These numbers
  say "the groups do not overlap", not "the difference matters". Read the mean differences
  and % columns for that.
- **The central practical result for RQ1.** S uses about half the energy of L. R saves only
  about 3% compared with L, while sending 66% of questions to the large model. Notebook 4
  looks at why.
- **η (correct answers per kJ) ranks R last.** L is about 17% more efficient than R, and S
  about 10% more than L. Routing gives up 15 correct answers while saving only 3% energy,
  so it ends up the least efficient strategy.
- **Several variables fail the normality check** and take the Kruskal-Wallis path. The log
  transform never fixes this.
  - Without the first run, routing's energy becomes normal (Shapiro p = 0.80), but S stays
    non-normal (p = 0.013), so the energy tests stay non-parametric.
  - Time and p95 latency are non-normal regardless.
  - Every test result and conclusion is the same with or without the first run.
- **The utilization variables are only "constant" because of one strategy.** For example,
  GPU 0 is unused under S and reads exactly 0%. The report's rule (non-zero variance in
  every group) then makes them descriptive. One could instead compare only the strategies
  that use that GPU; that's a team decision.
- **The tests assume independent runs.** The mild autocorrelation found in notebook 2 makes
  the p-values somewhat optimistic. Given their size, this doesn't change any conclusion,
  but it belongs in conclusion validity.
